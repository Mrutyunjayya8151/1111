```python
# Install BeautifulSoup if not already installed
!pip install beautifulsoup4
from bs4 import BeautifulSoup
# Sample HTML content (you can replace this with content from a file)
html_content = """
<html>
<head><title>Sample Page</title></head>
<body>
<h1>Welcome!</h1>
<p>This is an example of <b>HTML parsing</b> using BeautifulSoup.</p>
<a href="https://example.com">Click here</a>
</body>
</html>
"""
# Parse HTML and extract text
soup = BeautifulSoup(html_content, "html.parser")
text = soup.get_text()
# Output extracted text
print("Extracted Text:\n", text)
import nltk, spacy
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
# Downloads
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')  # we added this to resolve lookup error
spacy.cli.download("en_core_web_sm")
nlp = spacy.load("en_core_web_sm")
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
text = input("Enter your text:\n")
# Sentence tokenization
sentences = sent_tokenize(text)
# Word tokenization
words = word_tokenize(text)
# Stopword removal
filtered = [w for w in words if w.isalpha() and w.lower() not in stop_words]
# Stemming (stopwords removed)
stemmed = [stemmer.stem(w) for w in filtered]
# Lemmatization (stopwords removed)
lemmatized = [token.lemma_ for token in nlp(text) if token.is_alpha and token.text.lower() not
in stop_words]
```

```python
!pip install gensim
# Importing the Libraries
from gensim.models import Word2Vec
from gensim.utils import simple_preprocess

# Read dataset from a plain text file (one sentence per line)
with open("data_1000.txt", "r", errors="ignore") as f:
    sentences = [simple_preprocess(line) for line in f]

# Train Word2Vec
model = Word2Vec(sentences=sentences, vector_size=100, window=3, min_count=1, sg=1, epochs=200)

# Ask user for a word and show top-5 similar words
word = input("Enter a word: ").lower()

if word in model.wv:
    print(f"\nSimilar words to '{word}':")
    for w, score in model.wv.most_similar(word, topn=5):
        print(f"{w:15s} -> cosine: {score:.3f}")
else:
    print(f"'{word}' not found in vocabulary.")
```
